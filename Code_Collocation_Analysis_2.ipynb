```python
fileDir = "drive/MyDrive/corpus/coha_2000s.txt"
fr = open(fileDir, 'r', encoding='utf-8')
contents = fr.readlines()
fr.close()

textList = []
AllWordsList = []

num = 0
for content in contents:
    content = content.replace("\n", "")
    if num < 100:
        print(content)
    textList.append(content.lower())

    words = content.lower().strip().split(" ")
    for word in words:
        AllWordsList.append(word)
    num += 1
```

### Extract sentences that contain the word "pretty".

```python
wordsInCorpus = 0
targetWordText = []
AllTextWT = ""
for each in textList:
    words = each.split(" ")
    for word in words:
        wordsInCorpus += 1

    if " "+targetWord+" " in each:
        targetWordText.append(each)
        AllTextWT = AllTextWT + " " + each

AllText = AllTextWT.strip()
print(wordsInCorpus)
print(len(AllWordsList))
```

## Step 2: Concordance table

### Check the concordance table for sentences containing "pretty".

```python
from nltk import *
retokenize = RegexpTokenizer("[\w]+")
text = Text(retokenize.tokenize(AllText))
text.concordance(targetWord)
```

## Step 3: Create a contingency table

### Calculate co-occurrence frequencies and individual word frequencies.

```python
wordType = set()
for each in targetWordText:
    eachSplit = each.split(" ")
    for i in range(0,len(eachSplit)):
      if eachSplit[i] == targetWord and i != len(eachSplit)-1:
        wordType.add(eachSplit[i+1])
wordList = list(wordType)

print(wordList[0:10])
print(len(wordType))
```

### Calculate co-occurrence frequencies and filter words with a minimum frequency of 1.

```python
wordFreqDict = {}
for each in wordList:
    wordFreqDict[each] = 0

for each in targetWordText:
    eachSplit = each.split(" ")
    for i in range(0,len(eachSplit)):
      if eachSplit[i] == targetWord and i != len(eachSplit)-1:
        wordFreqDict[eachSplit[i+1]] = wordFreqDict.get(eachSplit[i+1]) + 1

wordFreqDicSorted = dict(sorted(wordFreqDict.items(), key=lambda x: x[1], reverse=True))

CoOccurringWords = []
collocations = []

topNum = 0
for key, value in wordFreqDicSorted.items():
  if topNum < 30:
    if value > 1:
        CoOccurringWords.append(key)
        collocations.append(value)
        print(key, value)
  topNum += 1

print(len(CoOccurringWords))
print(CoOccurringWords[0:10])
print(collocations[0:10])
```

### Calculate individual frequencies of co-occurring words.

```python
CoOccurringWordsFreqDict = {}
for each in CoOccurringWords:
    CoOccurringWordsFreqDict[each] = AllWordsList.count(each)
    print(each, AllWordsList.count(each))
```

```python
freqCoWord = []
freqTargetWord = []
targetword = []
wordsInCorpusList = []
targetwordCount = AllWordsList.count(targetWord)

for key, value in CoOccurringWordsFreqDict.items():
    freqCoWord.append(value)
    freqTargetWord.append(targetwordCount)
    targetword.append(targetWord)
    wordsInCorpusList.append(wordsInCorpus)

print(freqCoWord[0:10])
```

### Generate the final dataset.

```python
import pandas as pd
collocationTable = pd.DataFrame({'Target word':targetword,'Co-occurring words':CoOccurringWords,'Collocations':collocations,'Freq co-occurring word':freqCoWord,'Freq target word':freqTargetWord,'Words in corpus':wordsInCorpusList})
print(collocationTable)
```

## Step 4: Collocation measures
**Objective**: Apply the learned collocation measures to the dataset.

### Expected Frequency
The formula for Expected Frequency is:

$$E = \frac{f(w_1) \cdot f(w_2)}{N}$$

Where:
- $f(w_1)$: Frequency of the target word.
- $f(w_2)$: Frequency of the co-occurring word.
- $N$: Total number of tokens in the corpus.

```python
collocationTable['ExpectedFreq'] = (collocationTable['Freq target word']*collocationTable['Freq co-occurring word'])/collocationTable['Words in corpus']
```

```python
collocationTable[0:30]
```

```python
collocationTable.sort_values(by='ExpectedFreq', ascending=False)[0:30]
```

```python
collocationTable['OF-EF'] = collocationTable['Collocations']-collocationTable['ExpectedFreq']
```

```python
collocationTable.sort_values(by='OF-EF', ascending=False)[0:30]
```

### T-Score
The formula for T-Score is:

$$T = \frac{f(w_1, w_2) - E}{\sqrt{f(w_1, w_2)}}$$

Where:
- $f(w_1, w_2)$: Observed frequency of the collocation.
즉, 관찰빈도
- $E$: Expected frequency.

```python
import numpy as np
collocationTable['T-score'] = (collocationTable['Collocations'] - collocationTable['ExpectedFreq'])/np.sqrt(collocationTable['Collocations'])
collocationTable.sort_values(by='T-score', ascending=False)[0:30]
```

### Z-Score
The formula for Z-Score is:

$$Z = \frac{f(w_1, w_2) - E}{\sqrt{E}}$$

Where:
- $f(w_1, w_2)$: Observed frequency of the collocation.
- $E$: Expected frequency.

```python
import numpy as np
collocationTable['Z-score'] = (collocationTable['Collocations'] - collocationTable['ExpectedFreq'])/np.sqrt(collocationTable['ExpectedFreq'])
collocationTable.sort_values(by='Z-score', ascending=False)[0:30]
```

### Mutual Information (MI)
The formula for Mutual Information is:

$$MI = \log_2\left(\frac{f(w_1, w_2)}{N} \bigg/ \left(\frac{f(w_1)}{N} \cdot \frac{f(w_2)}{N}\right)\right)$$

Where:
- $f(w_1)$: Frequency of the target word.
- $f(w_2)$: Frequency of the co-occurring word.
- $N$: Total number of tokens in the corpus.
- $f(w_1, w_2)$: Observed frequency of the collocation.

```python
import numpy as np
collocationTable['MI'] = np.log2((collocationTable['Collocations']/collocationTable['Words in corpus'])/((collocationTable['Freq target word']/collocationTable['Words in corpus'])*(collocationTable['Freq co-occurring word']/collocationTable['Words in corpus'])))
collocationTable.sort_values(by='MI', ascending=False)[0:30]
```

### Log-Likelihood (LL)
The formula for Log-Likelihood is:

$$LL = 2 \cdot \sum \left(O \cdot \log\left(\frac{O}{E}\right)\right)$$

Where:
- $O$: Observed frequency.
- $E$: Expected frequency.

For collocation analysis, the expanded formula is:

$$LL = 2 \cdot \left(O_{11} \cdot \log\left(\frac{O_{11}}{E_{11}}\right) + O_{12} \cdot \log\left(\frac{O_{12}}{E_{12}}\right) + O_{21} \cdot \log\left(\frac{O_{21}}{E_{21}}\right) + O_{22} \cdot \log\left(\frac{O_{22}}{E_{22}}\right)\right)$$

Where:
- $O_{11}$: Observed frequency of the collocation.
- $O_{12}$: Observed frequency of the target word without the collocate.
- $O_{21}$: Observed frequency of the collocate without the target word.
- $O_{22}$: Observed frequency of all other words.
- $E_{11}, E_{12}, E_{21}, E_{22}$: Corresponding expected frequencies.

```python
def loglikelihood(score1, score2, score3, score4):
    outList = []
    for i in range(0,len(score1)):
        import math
        result = 2 * (
        score1[i] * math.log(score1[i], math.e) +
        score2[i] * math.log(score2[i], math.e) +
        score3[i] * math.log(score3[i], math.e) +
        score4[i] * math.log(score4[i], math.e) -
        (score1[i] + score2[i]) * math.log(score1[i] + score2[i], math.e) -
        (score1[i] + score3[i]) * math.log(score1[i] + score3[i], math.e) -
        (score2[i] + score4[i]) * math.log(score2[i] + score4[i], math.e) -
        (score3[i] + score4[i]) * math.log(score3[i] + score4[i], math.e) +
        (score1[i] + score2[i] + score3[i] + score4[i]) * math.log(score1[i] + score2[i] + score3[i] + score4[i], math.e))

        outList.append(result)
    return outList
```

```python
collocationTable['LL'] = loglikelihood(collocationTable['Collocations'],(collocationTable['Freq target word']-collocationTable['Collocations']),(collocationTable['Freq co-occurring word']-collocationTable['Collocations']),(collocationTable['Words in corpus']-collocationTable['Freq target word'])-(collocationTable['Freq co-occurring word']-collocationTable['Collocations']))
collocationTable.sort_values(by='LL', ascending=False)[0:30]
```
